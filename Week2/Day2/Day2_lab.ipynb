Roll two fair six-sided dice.

Event A = "at least one die shows a 6"
Event B = "the sum of the two dice is 8"
p(A|B)=P(A ∩ B) / P(B)
      =(2/36)/(5/36)
      =2/5
      =0.4

```python
np.random.seed()
die1=np.random.randint(1,7,100000)  # 1=if it shows 6 ,0 = other numbers
die2=np.random.randint(1,7,100000) 

event_A=(die1==6)|(die2==6)
event_B=(die1+die2)==8

prob=np.mean(event_A & event_B)/np.mean(event_B)
print(prob)
```
