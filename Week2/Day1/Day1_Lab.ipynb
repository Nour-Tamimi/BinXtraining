```python
age=insurance["age"].to_numpy()
```

```python
print(age)
```

```python
q1=np.percentile(age,25)
q3=np.percentile(age,75)
print("Mode",insurance["age"].mode()[0])

print("Mean",age.mean(),"\nMedian",np.median(age),"\nStandard Deviation",np.std(age),"\nVarience",np.var(age),"\nIQR",q3-q1)

```

Age is a numerical column i counted its mean, median, std, variance,mode and IQR 
The mean and median are closely the same but generally the median is better because its not affected by outliers
The std is like saying each value is in average of  14 away from the mean value 
The IQR is the range in the middle of the array of 50% of the sorted data
