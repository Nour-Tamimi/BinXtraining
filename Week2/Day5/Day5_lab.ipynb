```python
sns.histplot(med,x=med['age'],bins=30)
plt.title('Age distribution')
```

```python
q1=med['age'].quantile(0.25)
q3=med['age'].quantile(0.75)
iqr=q3-q1
c=med['age'].sort_index()

outliers = med[(c < q1 - 1.5*iqr) | (c > q3 + 1.5*iqr)]
print(outliers)
```

```python
med['age'].value_counts().sort_index()
```

here we can see the age dist from 18- 64 with average of 40 with high numbers in young ages 18-20 so its shape (right skewed) with no outlayers

```python
q1=med['bmi'].quantile(0.25)
q3=med['bmi'].quantile(0.75)
iqr=q3-q1
c=med['bmi'].sort_index()

outliers = med[(c < q1 - 1.5*iqr) | (c > q3 + 1.5*iqr)]
print(outliers)
print(outliers.count())
```
