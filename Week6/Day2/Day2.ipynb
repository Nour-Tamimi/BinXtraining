```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

## Plotting sigmoid, tanh and ReLU
over a range of random numbers to see what they do to input values

```python
#Generate inputs
input = np.linspace(-5, 5, 200)

#Compute the output
output_sigmoid = 1/(1 + np.exp(-input)) # sigmid 
output_tanh = np.tanh(input) # tanh
output_relu =np.maximum(0, input) #ReLU

#generate the plot
plt.figure(figsize=(10,6))

plt.plot(input, output_sigmoid, label='Sigmoid', color='blue',linewidth=2)
plt.plot(input, output_tanh, label='Tanh', color='red',linewidth=2)
plt.plot(input, output_relu, label='ReLU', color='green',linewidth=2)

plt.title("Activation Functions")
plt.xlabel("Input Values")
plt.ylabel("Output Values")

plt.axhline(0, color='black', linewidth=0.5, linestyle='--')
plt.axvline(0, color='black', linewidth=0.5, linestyle='--')

plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(fontsize=12)
plt.ylim(-1.5, 2.5)


plt.show()
```

**Sigmoid** makes all the output between 0 and 1 used for Output layer for binary classification as a probability.

**Tanh** makes all the output between -1 and 1 used for hidden layers when zero centered in valuable

**ReLU** converts negative values to zero, positive values kept the same, Used for hidden layers

## Best functions used for our heart-patients Project
for **hidden layers** we use **ReLU** as adefault since its fast and effective
for **output layer** we use **segmoid**, because our dataset is a binary classification (positive , negative)
for **loss function** we use **Binary cross-entropy**
 we use these for binary classification to calculate each class probability between 0 and 1


```python
# compute one forward pass for a tiny 2-layer network on a sample input.

# input layer 4 Neurons
input_1= [1.0, 2.0, 3.0, 2.5]

weights1 =[0.2, 0.8, -0.5, 1.0]
weights2 =[1.5, 2.0, 5,-0.7]

bias1 =2.0
bias2 =3.0


# 1 hidden layer 2 Neurons (dot product and activation ReLU)

v1= (input_1[0]*weights1[0] + input_1[1]*weights1[1] + input_1[2]*weights1[2] +input_1[3]*weights1[3]) +bias1
a1 =np.maximum(0,v1)

v2= (input_1[0]*weights2[0] + input_1[1]*weights2[1] + input_1[2]*weights2[2] +input_1[3]*weights2[3]) +bias2
a2 =np.maximum(0,v2)


#Output Layer(dot product and activation sigmoid), 1 Neurons
weight3= [0.1, 1]
bias3 =1.0

v3= (a1*weight3[0] + a2*weight3[1]) +bias3
a3 = 1 / (1 + np.exp(-v3))

print("The final Prediction", a3)

```
