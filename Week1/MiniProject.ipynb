```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
road=pd.read_csv('C:/Users/Lenovo/Downloads/automobile_dataset.csv')
```

```python
road.isnull().sum()
```

```python
road["Fuel_Efficiency"] = road["Fuel_Efficiency"].fillna(road["Fuel_Efficiency"].mean())
road["Accident_History"]=road["Accident_History"].fillna(road["Accident_History"].mean())
road["Transmission"]=road["Transmission"].fillna(road["Transmission"].mode()[0])
road.dropna(inplace=True)
road.drop_duplicates(inplace=True)
```

```python
milaege_array=road["Mileage"].to_numpy()
print("Milaege_array",milaege_array)
print("Mean",milaege_array.mean())
```

```python
plt.hist(road["Horsepower"]);
plt.title("Distribution of Horsepower")
```

```python
plt.scatter(road["Year"],road["Selling_Price"]);
plt.xlabel("Year")
plt.ylabel("Selling Price")
```

this is a scatter graph showing the relation between the selling price on the y axis and the year on the x axis

```python
plt.bar(road["Fuel_Type"],road["Mileage"]);
plt.xlabel("Fuel_Type")
plt.ylabel("Mileage")
plt.show()
```
