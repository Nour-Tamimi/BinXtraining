```python
#importing important libraries
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.model_selection import cross_val_score
from sklearn.linear_model import RidgeClassifier, Lasso
import pandas as pd
```

```python
X, y= load_breast_cancer(return_X_y=True)
scalar= StandardScaler()
X_scaled= scalar.fit_transform(X)

X_temp, X_test, y_temp, y_test =train_test_split(X_scaled, y, test_size=0.2, random_state=20)
X_train, X_val, y_train, y_val = train_test_split(X_temp, y_temp, test_size=0.7, random_state=20 )
```

```python
#OverFitting
model =DecisionTreeClassifier(max_depth=None,random_state=32,min_samples_split=2,min_samples_leaf=1)
model.fit(X_train,y_train)
predictions =model.predict(X_val)
train_predict=model.predict(X_train)
train_acc=accuracy_score(y_train,train_predict)
print("Train Accuracy",train_acc)

Val_acc =accuracy_score(y_val, predictions)
print("Validation Accuracy",Val_acc)
```

This is overfitting: there is a large gap between train and validation accuracy. Train accuracy is perfect (1.0), meaning the model has memorized the training data rather than learned a generalizable pattern. Validation accuracy (0.88) is noticeably lower, confirming the model does not generalize well to unseen data

```python
#UnderFitting
model2 =DecisionTreeClassifier(max_depth=1,              
    min_samples_leaf=200,     
    min_samples_split=300,   
    max_leaf_nodes=2,        
    random_state=32
)
model2.fit(X_train,y_train)

train_predict2=model2.predict(X_train)
train_acc2=accuracy_score(y_train,train_predict2)
print("Train Accuracy",train_acc2)

predictions2 =model2.predict(X_val)
Val_acc2 =accuracy_score(y_val, predictions2)
print("Validation Accuracy",Val_acc2)
```

This is underfitting since both scores are low train is (0.54), and validation is (0.68) so the model is weak on training and testing, and the gap between them is small. This is different from a good fit, the model must also be too simple to capture the pattern in the data 

```python
#regularization
ridge = RidgeClassifier(alpha=1.0)
ridge.fit(X_train,y_train)

train_acc_ridge = accuracy_score(y_train, ridge.predict(X_train))
val_acc_ridge = accuracy_score(y_val, ridge.predict(X_val))

print("Ridge Train accuracy:", train_acc_ridge)
print("Ridge Val accuracy:", val_acc_ridge)
```

after regulization and adding a penalty using ridge this make the feature shrink the model coofecient, This reduces the model's flexibility to fit noise in the training data towards zero so we get rid of overfitting
